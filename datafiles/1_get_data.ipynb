# Get data from server

Loads datafiles with batch, microscopy, and FACS data from [Github Repo](https://github.com/simonvanvliet/Spatial-Tolerance-Figure-Data-and-Code/)


```python
from pathlib import Path
import shutil
import urllib.request
```

## Copy from local repo

```python
path_data_repo = Path('/Users/simonvanvliet/Library/CloudStorage/Dropbox/Work/Code/Spatial-Tolerance-Figure-Data-and-Code')
repo_path = Path.cwd() 

file_list = [
    'Figure3/Figure3A-S12_CallibrationData/analysis_code/hqno_calibration_microscopy.csv',
    'Figure3/Figure3A-S12_CallibrationData/figure_code/hqno_calibration_facs.csv',
    'Figure3/Figure3A-S12_CallibrationData/analysis_code/rhl_calibration_microscopy.csv',
    'Figure3/Figure3A-S12_CallibrationData/figure_code/hqno_calibration_facs.csv',
    'Figure3/Figure3A-S12_CallibrationData/figure_code/rhl_calibration_facs.csv',
    'Figure3/Figure3C_SurvivalData/HQNO_RHL_Survival.csv',
    'Figure3/Figure3A-S11_ConcentrationGradients/analysis_code/hqno_gradient_microscopy.csv',
    'Figure3/Figure3A-S11_ConcentrationGradients/analysis_code/rhl_gradient_microscopy.csv' 
    ]

for file_path in file_list:
    src = path_data_repo / file_path
    file_name = Path(file_path).name
    dst = repo_path / file_name  # Explicitly use just the filename

    if src.exists():
        shutil.copy(src, dst)
    else:
        print(f"Server path does not exist: {src}")
```

## Download directly from Github repo

```python

GITHUB_RAW_BASE = 'https://raw.githubusercontent.com/simonvanvliet/Spatial-Tolerance-Figure-Data-and-Code/main'
repo_path = Path.cwd()

file_list = [
    'Figure3/Figure3A-S12_CallibrationData/analysis_code/hqno_calibration_microscopy.csv',
    'HQNO_Calibration/hqno_calibration_facs.csv',
    'Figure3/Figure3A-S12_CallibrationData/analysis_code/rhl_calibration_microscopy.csv',
    'Figure3/Figure3A-S12_CallibrationData/figure_code/hqno_calibration_facs.csv',
    'Figure3/Figure3A-S12_CallibrationData/figure_code/rhl_calibration_facs.csv',
    'Figure3/Figure3C_SurvivalData/HQNO_RHL_Survival.csv',
    'Figure3/Figure3A-S11_ConcentrationGradients/analysis_code/hqno_gradient_microscopy.csv',
    'Figure3/Figure3A-S11_ConcentrationGradients/analysis_code/rhl_gradient_microscopy.csv',
]

for file_path in file_list:
    url = f'{GITHUB_RAW_BASE}/{file_path}'
    dst = repo_path / Path(file_path).name
    try:
        urllib.request.urlretrieve(url, dst)
        print(f'Downloaded: {dst.name}')
    except Exception as e:
        print(f'Failed to download {file_path}: {e}')
```
